```python
# 1. อายุเฉลี่ยของผู้ป่วยหญิงทั้งหมด
from IPython.terminal.interactiveshell import Type
sum = 0
N = 0
for a in female['age']:
  try:
    a_num = float(a)
    if a > 0:
        sum += a # sum = sum + a
        N += 1
  except(ValueError, TypeError):
    continue
print(f'อายุเฉลี่ย ของ ผู้ป่วยหญิง {sum/N}')
```

```python
# อายุเฉลี่ยของผู้ป่วยชายทั้งหมด
from IPython.terminal.interactiveshell import Type
sum = 0
N = 0
for a in male['age']:
  try:
    a_num = float(a)
    if a > 0:
        sum += a # sum = sum + a
        N += 1
  except(ValueError, TypeError):
    continue
print(f'อายุเฉลี่ย ของ ผู้ป่วยชาย {sum/N}')
```

```python
# อายุเฉลี่ยของผู้ป่วยหญิงในขอนแก่น
female_KK = female[female['province_of_onset']=='ขอนแก่น']

sum = 0
N = 0
for a in female_KK['age']:
  try:
    a_num = float(a)
    if a > 0:
        sum += a # sum = sum + a
        N += 1
  except(ValueError, TypeError):
    continue
print(f'อายุเฉลี่ย ของ ผู้ป่วยหญิง ในจังหวัดขอนแก่น {sum/N}')
```

```python
# อายุเฉลี่ยของผู้ป่วยชายในขอนแก่น
male_KK = male[male['province_of_onset']=='ขอนแก่น']

sum = 0
N = 0
for a in male_KK['age']:
  try:
    a_num = float(a)
    if a > 0:
        sum += a # sum = sum + a
        N += 1
  except(ValueError, TypeError):
    continue
print(f'อายุเฉลี่ย ของ ผู้ป่วยชาย ในจังหวัดขอนแก่น {sum/N}')
```
